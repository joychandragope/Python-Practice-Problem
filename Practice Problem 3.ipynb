```python
def double(x):
    return x*2

print(double(5))
```

```python
double = lambda x: x*2
print(double(5))
```

```python
avg = lambda x,y,z: (x+y+z)/2
print(avg(2,4,6))
```

```python
cube = lambda x: x*x*x
print(cube(5))
```

```python
def cube(x):
    return x * x * x

l = [1,2,3,4,5,6]
new_list = list(map(cube,l))
print(new_list)
```

### Given a list [1,2,3,4], use map() and lambda to create a new list with squares of each number.

```python
data_list = [1,2,3,4]

#def square_list(x):
  #  return x * x

# Using Map
new_list = list(map(lambda x: x*x, data_list))
print(new_list)
```
